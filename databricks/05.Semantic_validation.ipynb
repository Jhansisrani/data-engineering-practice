```python
#Step 1 — Create intentionally bad data
from datetime import date

semantic_data = [
    ("T1501", "C1501", "P1501", "S1501",
     date(2026, 9, 16), 2, -50.0, -100.0, "PROMO10", "WEB")
]
semantic_columns = [
    "transaction_id",
    "customer_id",
    "product_id",
    "store_id",
    "transaction_date",
    "quantity",
    "unit_price",
    "total_amount",
    "promo_code",
    "source_system"
]
semantic_df = spark.createDataFrame(semantic_data, semantic_columns)
semantic_df.printSchema()
display(semantic_df)
```

```python
display(semantic_df.show())
```

```python
semantic_df.filter("unit_price <= 0").show()
```

```python
#Rule 1 — Quantity must be greater than 0
semantic_df.filter("quantity <= 0").show()
```

```python
#Rule 2 — Unit price must be greater than 0
semantic_df.filter("unit_price <= 0").show()
```

```python
#Rule 3 — Total amount must be >= 0
semantic_df.filter("total_amount < 0").show()
```

```python
from pyspark.sql.functions import when, col

validated_df = semantic_df.withColumn( "validation_status",
                                      when(
        (col("quantity") > 0) &
        (col("unit_price") > 0) &
        (col("total_amount") >= 0),
        "VALID"
    ).otherwise("INVALID")
)

display(validated_df)
```
