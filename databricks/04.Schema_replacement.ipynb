```python
 

from datetime import date

new_data = [
    ("T1401", "C1401", "P1401", "S1401",
     date(2026, 9, 15), 1, 75.0, 75.0,
     "PROMO20", "APP", "UPI")
]

new_columns = [
    "transaction_id",
    "customer_id",
    "product_id",
    "store_id",
    "transaction_date",
    "quantity",
    "unit_price",
    "total_amount",
    "promo_code",
    "source_system",
    "payment_method"
]

overwrite_df = spark.createDataFrame(new_data, new_columns)

overwrite_df.printSchema()
display(overwrite_df)
```

```python
overwrite_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.my_databricks_demo.transactions")
```
