```python
# Step 03 - Define Auto Loader stream

catalog = "retail"
schema = "sales"
volume = "autoloader_demo"

base = f"/Volumes/{catalog}/{schema}/{volume}/auto_loader"

landing = f"{base}/landing"
schema_loc = f"{base}/_schema"

# Define the Auto Loader streaming DataFrame
orders_stream = (
    spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "json")
    .option("cloudFiles.schemaLocation", schema_loc)
    .load(landing)
)

print("Auto Loader stream defined.")
```

```python
# Step 04 - Start Auto Loader stream

"""catalog = "retail"
schema = "sales"
volume = "autoloader_demo"

base = f"/Volumes/{catalog}/{schema}/{volume}/auto_loader") """

checkpoint = f"{base}/_checkpoint"

table_name = f"{catalog}.{schema}.orders_bronze"

# Start the streaming query
query = (
    orders_stream.writeStream
    .option("checkpointLocation", checkpoint)
      .trigger(availableNow=True)
    .toTable(table_name)
)

print("Streaming query started.")
```
