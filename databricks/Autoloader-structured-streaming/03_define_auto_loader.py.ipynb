```python
# Step 03 - Define Auto Loader stream

catalog = "retail"
schema = "sales"
volume = "autoloader_demo"

base = f"/Volumes/{catalog}/{schema}/{volume}/auto_loader"

landing = f"{base}/landing"
schema_loc = f"{base}/_schema"

# Define the Auto Loader streaming DataFrame
orders_stream = (
    spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "json")
    .option("cloudFiles.schemaLocation", schema_loc)
    .load(landing)
)

print("Auto Loader stream defined.")
```
