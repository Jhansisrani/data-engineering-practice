```python
# ============================================================
# MODULE 3 - SCHEMA EVOLUTION
# Step 1 - Create incoming data with a new column
# ============================================================

from datetime import date

data = [
    ("T1301", "C1301", "P1301", "S1301", date(2026, 9, 14), 2, 50.0, 100.0, "PROMO10", "WEB")
]

columns = [
    "transaction_id",
    "customer_id",
    "product_id",
    "store_id",
    "transaction_date",
    "quantity",
    "unit_price",
    "total_amount",
    "promo_code",
    "source_system"
]

evolution_df = spark.createDataFrame(data, columns)

evolution_df.printSchema()
display(evolution_df)
```

```python
# ============================================================
# MODULE 3 - SCHEMA EVOLUTION
# Step 2 - Append new column without schema evolution
# ============================================================

table_name = "workspace.my_databricks_demo.transactions"

evolution_df.write.format("delta").mode("append").saveAsTable(table_name)
```

```python
# ============================================================
# MODULE 3 - SCHEMA EVOLUTION
# Step 3 - Append with schema evolution
# ============================================================

evolution_df.write.format("delta").mode("append").option("mergeSchema", "true").saveAsTable(table_name)
```

```python
spark.sql("SELECT * FROM workspace.my_databricks_demo.transactions").show()
```
