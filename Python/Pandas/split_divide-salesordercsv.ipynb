```python
#slcing with iloc

df2.iloc[:2]       # first 2 rows, all columns
df2.iloc[:5]       # first 5 rows, all columns
df2.iloc[2:5]      # rows 3, 4, 5, all columns
df2.iloc[:, :2]    # all rows, first 2 columns
df2.iloc[0, 0]     # first row, first column, one value
```

```python
df2.loc[:1]      # index 0 and 1 → 2 rows
df2.iloc[:2]     # position 0 and 1 → 2 rows
```

```python
df2.loc[:2]      # index 0, 1, 2 → 3 rows
df2.iloc[:2]     # position 0, 1 → 2 rows
```

Yes 👍 That's the main difference.

.loc → labels

You can pass strings because .loc works with row/column labels.

df3.loc[df3.index[0], "SALESORDERID"]

Here "SALESORDERID" is the column label.

You can also use a string row label if your DataFrame index contains strings:

df.loc["customer_101", "SALESORDERID"]
.iloc → positions

.iloc works with integer positions only:

df3.iloc[0, 0]

Means:

1st row, 1st column.

You cannot do:

df3.iloc[0, "SALESORDERID"]  # ❌

because "SALESORDERID" is a label, not a position.

Easy memory trick

loc = Label

loc → "SALESORDERID"

iloc = Integer Location

iloc → 0, 1, 2...


```python
df3.loc[df3.index[0], "SALESORDERID"]   # label-based
df3.iloc[0, 0]                          # position-based
```

```python
# Missing Sales Order ID
df3.loc[df3.index[0], "SALESORDERID"] = ""

# Missing Partner ID
df3.loc[df3.index[1], "PARTNERID"] = ""

# Empty Gross Amount
df3.loc[df3.index[2], "GROSSAMOUNT"] = ""

# Empty Net Amount
df3.loc[df3.index[3], "NETAMOUNT"] = ""

# Negative Gross Amount
df3.loc[df3.index[4], "GROSSAMOUNT"] = "-500"

# Negative Net Amount
df3.loc[df3.index[5], "NETAMOUNT"] = "-250"
```

```python
# Missing SALESORDERID
df3.iloc[0, df3.columns.get_loc("SALESORDERID")] = ""

# Missing PARTNERID
df3.iloc[1, df3.columns.get_loc("PARTNERID")] = ""

# Empty GROSSAMOUNT
df3.iloc[2, df3.columns.get_loc("GROSSAMOUNT")] = ""

# Empty NETAMOUNT
df3.iloc[3, df3.columns.get_loc("NETAMOUNT")] = ""

# Negative GROSSAMOUNT
df3.iloc[4, df3.columns.get_loc("GROSSAMOUNT")] = "-500"

# Negative NETAMOUNT
df3.iloc[5, df3.columns.get_loc("NETAMOUNT")] = "-250"

# Duplicate SALESORDERID
df3.iloc[6, df3.columns.get_loc("SALESORDERID")] = df3.iloc[7]["SALESORDERID"]
```

```python
# First 25 → missing SALESORDERID
for i in range(0, 25):
    df3.loc[df3.index[i], "SALESORDERID"] = ""

# Next 25 → missing PARTNERID
for i in range(25, 50):
    df3.loc[df3.index[i], "PARTNERID"] = ""

# Next 25 → negative GROSSAMOUNT
for i in range(50, 75):
    df3.loc[df3.index[i], "GROSSAMOUNT"] = "-500"

# Next 25 → negative NETAMOUNT
for i in range(75, 100):
    df3.loc[df3.index[i], "NETAMOUNT"] = "-250"
```
