```python
#Write a Python function to find the union of two sets without using the union() method or | operator.

set1={1,2,3,4,5}
set2={1,2,3,4,5,6,7,8}

def fn_union (set1,set2):
   res=set()
   for i in set1:
    res.add(i)
    for j in set2:      
     res.add(j)
   return(res)    
```

```python
#Write a Python function to find the symmetric difference of two sets without using symmetric_difference() or ^ operator.

set1={1,2,3,4,5}
set2={1,2,3,4,5,6,7,8}

def fn_symdiff (set1,set2):
   res=set()
   for i in set1:
    for j in set2: # this will check every loop,so will try another logic
       if j == i:
        res.clear()
       else:
        res.add(j)
   return(res) 
```

```python
fn_symdiff (set1,set2)
```

```python
#symmetric difference using nested for loop without function
set1={1,2,3,4,5}
set2={1,2,3,4,5,6,7,8}

def fn_symdiff (set1,set2):
   res=set()
   for i in set1:
    if i not in set2:
     res.add(i)
   for j in set2:  
     if j not in set1:
      res.add(j)
   return(res) 
```
