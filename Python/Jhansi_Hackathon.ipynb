```python
#using break and continue

'''🔹 Scenario: Bank Transactions

You are processing a list of withdrawal requests for a customer.

Rules:

If the requested withdrawal amount is more than the balance + overdraft, skip that transaction (use continue)

If the requested withdrawal amount is negative, stop processing all transactions (use break)

Otherwise, process the withdrawal and print the new balance'''
```

```python
balance=5000
overdraft=1000

withdrawals = [2000, 7000, -500, 3000, 1000]

for i in withdrawals:
  amt=balance+overdraft
 
  if i>amt:
    print("cannot withdraw,skipping",i)
    continue
  elif i<0:
    print("invalid transaction amount,exit",i)
    break
  else:
    new_bal=balance-i
  print("balance is",new_bal,"after withdraw",i)
```
